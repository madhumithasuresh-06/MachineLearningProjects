```python
data_yaml = """
path: /content/dataset/dataset
train: images
val: images

nc: 6
names: ["circle","square","triangle","rectangle","pentagon","hexagon"]
"""

with open("dataset.yaml", "w") as f:
    f.write(data_yaml)
```

```python
from ultralytics import YOLO

model = YOLO("yolov8n.pt")
model.train(data="dataset.yaml", epochs=20)
```

```python
model.val()
```

```python
classes = ["circle","square","triangle","rectangle","pentagon","hexagon"]
```

```python
EXPECTED_POSITIONS = {
    "circle": (0.15, 0.25),
    "square": (0.5, 0.25),
    "triangle": (0.85, 0.25),
    "rectangle": (0.2, 0.75),
    "pentagon": (0.5, 0.75),
    "hexagon": (0.85, 0.75)
}
```

```python
def analyze_image(image_path):
    img = cv2.imread(image_path)
    h, w, _ = img.shape

    results = model(image_path)

    detected = []

    # Draw detected objects (GREEN)
    for r in results:
        for box in r.boxes:
            cls_id = int(box.cls[0])
            cls_name = classes[cls_id]
            detected.append(cls_name)

            x1, y1, x2, y2 = map(int, box.xyxy[0])

            cv2.rectangle(img, (x1,y1), (x2,y2), (0,255,0), 2)
            cv2.putText(img, cls_name, (x1,y1-10),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0,255,0), 2)

    # Find missing
    missing = list(set(classes) - set(detected))

    # Draw RED boxes for missing objects
    for m in missing:
        cx, cy = EXPECTED_POSITIONS[m]

        x = int(cx * w)
        y = int(cy * h)

        box_size = 80

        x1 = x - box_size
        y1 = y - box_size
        x2 = x + box_size
        y2 = y + box_size

        cv2.rectangle(img, (x1,y1), (x2,y2), (0,0,255), 3)
        cv2.putText(img, f"Missing: {m}", (x1,y1-10),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0,0,255), 2)

    print("Detected:", detected)
    print("Missing:", missing)

    # Show image in Colab
    from google.colab.patches import cv2_imshow
    cv2_imshow(img)

    return detected, missing
```

```python
analyze_image("/content/dataset/dataset/images/train/image_0006.jpg")
```
